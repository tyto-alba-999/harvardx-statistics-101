### Jar Example: Sampling with Replacement

Imagine a jar containing $n$ balls labeled from $1$ to $n$. We draw one ball at a time, returning each ball to the jar before the next draw.

Since the ball is replaced after every draw, each draw has $n$ possible outcomes. For a sample of size $k$, there are $k$ successive draws.

By the multiplication rule, the total number of possible ordered samples is

$$
|\Omega| = \underbrace{n \times n \times \cdots \times n}_{k\text{ times}} = n^k.
$$

**Key idea:** Because sampling is performed with replacement, the same ball can be selected multiple times, and every draw has the same $n$ possible outcomes.

```python
from itertools import product

# Parameters
n = 3  # Number of balls
k = 2  # Number of draws

balls = list(range(1, n + 1))

# Sampling with replacement: each draw can select any ball
sample_space = list(product(balls, repeat=k))

print(f"Balls: {balls}")
print(f"Number of draws: {k}\n")

print("Sample space:")
for sample in sample_space:
    print(sample)

# Verify the multiplication rule
print(f"\nNumber of possible samples: {len(sample_space)}")
print(f"Multiplication rule: {n}^{k} = {n**k}")

assert len(sample_space) == n**k
```
