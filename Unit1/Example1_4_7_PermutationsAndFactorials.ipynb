### Example 1.4.7: Permutations and Factorials

A **permutation** of $1, 2, \ldots, n$ is an arrangement of these numbers in a particular order.

For example, $(3, 5, 1, 2, 4)$ is a permutation of $1, 2, 3, 4, 5$.

To construct a permutation, we choose the numbers one at a time without replacement:

- The first position has $n$ possible choices.
- The second position has $n-1$ remaining choices.
- The third position has $n-2$ remaining choices.
- This continues until only one number remains.

By the multiplication rule, the total number of permutations is

$$
n(n-1)(n-2)\cdots 2 \cdot 1 = n!.
$$

The **factorial** of a positive integer $n$ is defined as

$$
n! = \prod_{j=1}^{n} j.
$$

By convention, $0! = 1$.

For example, the number of ways to arrange $5$ distinct people in a line is

$$
5! = 5 \times 4 \times 3 \times 2 \times 1 = 120.
$$

```python
from itertools import permutations
from math import factorial

# Numbers to arrange
n = 4
numbers = list(range(1, n + 1))

# Generate all permutations
all_permutations = list(permutations(numbers))

print(f"Numbers: {numbers}")
print(f"Total permutations: {len(all_permutations)}")
print(f"Factorial calculation: {n}! = {factorial(n)}")

print("\nFirst 10 permutations:")
for permutation in all_permutations[:10]:
    print(permutation)

# Verify the multiplication rule
assert len(all_permutations) == factorial(n)
```

```python
from itertools import permutations
from math import factorial

print("n | Multiplication calculation | n! | Number of permutations")
print("-" * 65)

for n in range(1, 7):
    numbers = list(range(1, n + 1))
    count = sum(1 for _ in permutations(numbers))

    multiplication = " × ".join(map(str, range(n, 0, -1)))

    print(f"{n} | {multiplication:<27} | {factorial(n):<3} | {count}")

    assert count == factorial(n)

# Verify the special case
print(f"\n0! = {factorial(0)}")
assert factorial(0) == 1
```

### Conclusion

A permutation is an arrangement in which the order matters and each element is used exactly once.

The multiplication rule gives $n!$ possible arrangements of $n$ distinct elements:

$$
n! = n(n-1)(n-2)\cdots 1.
$$

The factorial function also satisfies the convention $0! = 1$, which is useful in many counting formulas.
