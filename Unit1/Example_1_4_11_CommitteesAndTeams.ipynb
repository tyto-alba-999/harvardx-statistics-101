
### Example 1.4.11 (Committees and teams)

Consider a group of four people.

* **(a)** How many ways are there to choose a two-person committee?
* **(b)** How many ways are there to break the people into two teams of two?

---

### Solution

#### **(a)** Choosing a two-person committee

One way to count the possibilities is by directly listing them out. Labeling the people as $1, 2, 3, 4$, the possible committees are:

$$
\{1, 2\}, \{1, 3\}, \{1, 4\}, \{2, 3\}, \{2, 4\}, \{3, 4\}
$$

*(This gives a total of 6 possibilities).*

Another approach is to use the **multiplication rule** combined with an **adjustment for overcounting**:

* There are $4$ ways to choose the first person on the committee.
* There are $3$ ways to choose the second person.

However, this ordered process counts each committee twice. For example, picking person $1$ then person $2$ results in the exact same committee as picking person $2$ then person $1$ ($\{1, 2\} = \{2, 1\}$). Since we have overcounted by a factor of $2$, we divide by $2$ to get the correct count:

$$
\frac{4 \times 3}{2} = 6
$$

---

#### **(b)** Breaking into two teams of two

To find the number of ways to break the four people into two separate teams of two, let's look at a few ways to think about it:

1. **Direct Listing:** Labeling the people as $1, 2, 3, 4$, we can write out the possible pairings:

   * Team 1: $\{1, 2\}$ and Team 2: $\{3, 4\}$
   * Team 1: $\{1, 3\}$ and Team 2: $\{2, 4\}$
   * Team 1: $\{1, 4\}$ and Team 2: $\{2, 3\}$

   *(This gives a total of 3 distinct ways).*

2. **Fixing a Teammate:** Alternatively, you can simply specify who is on person $1$'s team (which automatically determines the other team). Once person $1$ is paired with one of the remaining $3$ people, the other two people automatically form the second team.

3. **Using Part (a) with Adjustment:** We can use our result from part (a), which showed there are $6$ ways to choose a single two-person committee. However, choosing one team of two automatically leaves behind the second team of two. This overcounts the overall division into *two unnamed teams* by a factor of $2$, because picking team $\{1, 2\}$ (leaving $\{3, 4\}$) is the same partition as picking team $\{3, 4\}$ (leaving $\{1, 2\}$). Adjusting for this overcounting gives:

$$
\frac{6}{2} = 3
$$


```python

from itertools import combinations

# Four distinct people
people = (1, 2, 3, 4)

# --------------------------------------------------
# Part (a): Choosing a two-person committee
# --------------------------------------------------

# Method 1: Direct listing
committees = list(combinations(people, 2))

print("PART (a): TWO-PERSON COMMITTEES")
print("-" * 38)

for committee in committees:
    print(committee)

print(f"\nTotal committees: {len(committees)}")

# Method 2: Multiplication rule with overcounting
ordered_selections = [
    (first, second)
    for first in people
    for second in people
    if first != second
]

print(f"Ordered selections: {len(ordered_selections)}")
print(f"After correcting for overcounting: "
      f"{len(ordered_selections)} / 2 = "
      f"{len(ordered_selections) // 2}")


# --------------------------------------------------
# Part (b): Dividing into two unnamed teams of two
# --------------------------------------------------

print("\nPART (b): DIVISIONS INTO TWO TEAMS")
print("-" * 38)

# Method 1: Direct listing
divisions = []

for team1 in combinations(people, 2):
    team2 = tuple(p for p in people if p not in team1)

    # Keep only divisions where person 1 belongs to team 1.
    # This avoids counting a division twice.
    if 1 in team1:
        divisions.append((team1, team2))

print("\nMethod 1 — Direct listing:")
for team1, team2 in divisions:
    print(f"Team 1: {team1} | Team 2: {team2}")

print(f"Total divisions: {len(divisions)}")

# Method 2: Fixing person 1's teammate
print("\nMethod 2 — Fixing person 1's teammate:")
for teammate in people[1:]:
    team1 = (1, teammate)
    team2 = tuple(p for p in people if p not in team1)
    print(f"Person 1 with {teammate}: "
          f"{team1} | {team2}")

# Method 3: Using part (a) with adjustment
print("\nMethod 3 — Correcting for overcounting:")
print(f"{len(committees)} / 2 = "
      f"{len(committees) // 2}")

# Verify the results
assert len(committees) == 6
assert len(divisions) == 3
```

```python

print("COMMITTEES")
print("=" * 42)
print("Each bracketed pair is one possible committee.\n")

for i, committee in enumerate(committees, start=1):
    selected = set(committee)
    representation = "  ".join(
        f"[{person}]" if person in selected else f" {person} "
        for person in people
    )
    print(f"{i}. {representation}")

print("\nTEAM DIVISIONS")
print("=" * 42)
print("Each line represents one complete division.\n")

for i, (team1, team2) in enumerate(divisions, start=1):
    print(f"Division {i}")
    print(f"    Team 1: {team1}")
    print(f"    Team 2: {team2}")
    print("    " + "-" * 24)

print("FINAL COUNT")
print("-" * 24)
print(f"Two-person committees: {len(committees)}")
print(f"Divisions into two teams: {len(divisions)}")
```
