### Example 1.4.16 (Full house in poker).

A 5-card hand is dealt from a [standard, well-shuffled 52-card deck](https://www.thoughtco.com/standard-deck-of-cards-3126599). The hand is called a *full house* in poker if it consists of three cards of some rank and two cards of another rank, e.g., three 7's and two 10's (in any order). What is the probability of a full house?

### Solution

All of the $\binom{52}{5}$ possible hands are equally likely by symmetry, so the naive definition is applicable. To find the number of full house hands, use the multiplication rule (and imagine the tree). There are 13 choices for what rank we have three of; for concreteness, assume we have three 7's and focus on that branch of the tree. There are $\binom{4}{3}$ ways to choose which 7's we have. Then there are 12 choices for what rank we have two of, say 10's for concreteness, and $\binom{4}{2}$ ways to choose two 10's. Thus,

$$P(\text{full house}) = \frac{13\binom{4}{3}12\binom{4}{2}}{\binom{52}{5}} = \frac{3744}{2598960} \approx 0.00144.$$

The decimal approximation is more useful when playing poker, but the answer in terms of binomial coefficients is exact and *self-annotating* (seeing "$\binom{52}{5}$" is a much bigger hint of its origin than seeing "2598960").

### Understanding the Counting Process

A full house consists of three cards of one rank and two cards of a different rank.

The numerator counts the favorable hands:

1. Choose the rank that appears three times.
2. Choose three of its four suits.
3. Choose a different rank for the pair.
4. Choose two of its four suits.

The denominator counts every possible 5-card hand.

Since every 5-card hand is equally likely, the probability is the ratio of favorable hands to all possible hands.

```python
from math import comb


# Standard deck parameters
number_of_ranks = 13
suits_per_rank = 4
hand_size = 5

three_of_a_kind = 3
pair_size = 2

# ============================================================
# 1. COUNT ALL POSSIBLE HANDS
# ============================================================

total_cards = number_of_ranks * suits_per_rank

total_hands = comb(total_cards, hand_size)

# ============================================================
# 2. COUNT FAVORABLE HANDS
# ============================================================

# Choose the rank for the three-of-a-kind
choose_triple_rank = number_of_ranks

# Choose 3 suits out of 4
choose_three_cards = comb(suits_per_rank, three_of_a_kind)

# Choose a DIFFERENT rank for the pair
choose_pair_rank = number_of_ranks - 1

# Choose 2 suits out of 4
choose_two_cards = comb(suits_per_rank, pair_size)

# Multiplication rule
full_house_hands = (
    choose_triple_rank
    * choose_three_cards
    * choose_pair_rank
    * choose_two_cards
)

# ============================================================
# 3. CALCULATE THE PROBABILITY
# ============================================================

probability = full_house_hands / total_hands
percentage = probability * 100

# ============================================================
# 4. DISPLAY RESULTS
# ============================================================

print("=" * 70)
print("FULL HOUSE IN POKER")
print("=" * 70)

print("\nDECK AND HAND")
print("-" * 70)
print(f"Number of ranks: {number_of_ranks}")
print(f"Number of suits per rank: {suits_per_rank}")
print(f"Total number of cards: {number_of_ranks} x {suits_per_rank} = {total_cards}")
print(f"Number of cards dealt: {hand_size}")

print("\nALL POSSIBLE HANDS")
print("-" * 70)
print(f"C({total_cards}, {hand_size})")
print(f"= {total_cards}! / ({hand_size}! x {total_cards - hand_size}!)")
print(f"= {total_hands:,}")

print("\nFAVORABLE HANDS — MULTIPLICATION RULE")
print("-" * 70)

print(f"1. Choose the rank for the three-of-a-kind: {number_of_ranks}")
print(
    f"2. Choose {three_of_a_kind} suits out of {suits_per_rank}: "
    f"C({suits_per_rank}, {three_of_a_kind}) = {choose_three_cards}"
)
print(f"3. Choose a different rank for the pair: {number_of_ranks - 1}")
print(
    f"4. Choose {pair_size} suits out of {suits_per_rank}: "
    f"C({suits_per_rank}, {pair_size}) = {choose_two_cards}"
)

print("\nMultiplication:")
print(
    f"{number_of_ranks} x {choose_three_cards} x "
    f"{number_of_ranks - 1} x {choose_two_cards}"
)
print(f"= {full_house_hands:,} favorable hands")

print("\nPROBABILITY")
print("-" * 70)
print(f"P(full house) = Favorable hands / All possible hands")
print(f"              = {full_house_hands:,} / {total_hands:,}")
print(f"              = {probability:.10f}")
print(f"              ≈ {probability:.5f}")
print(f"              = {percentage:.6f}%")

print("\nVERIFICATION")
print("-" * 70)
print(f"Expected favorable hands: 3,744")
print(f"Calculated favorable hands: {full_house_hands:,}")
print(f"Expected total hands: 2,598,960")
print(f"Calculated total hands: {total_hands:,}")

passed = (
    full_house_hands == 3744
    and total_hands == 2598960
)

print("\nResult:", "PASSED" if passed else "FAILED")

print("\n" + "=" * 70)
```

```python
# ============================================================
# FULL HOUSE COUNTING TREE
# ============================================================

from math import comb

print("=" * 70)
print("COUNTING TREE — FULL HOUSE")
print("=" * 70)

print("""
START
  |
  |-- Choose the rank for THREE cards
  |      13 possible ranks
  |
  |-- Choose THREE suits from that rank
  |      C(4, 3) = 4 possibilities
  |
  |-- Choose a DIFFERENT rank for TWO cards
  |      12 remaining ranks
  |
  |-- Choose TWO suits from that rank
         C(4, 2) = 6 possibilities
""")

print("NUMBER OF BRANCHES AT EACH STAGE")
print("-" * 70)

stages = [
    ("Choose the triple rank", 13),
    ("Choose its three suits", comb(4, 3)),
    ("Choose a different pair rank", 12),
    ("Choose its two suits", comb(4, 2))
]

running_total = 1

for description, choices in stages:
    running_total *= choices
    print(f"{description:<38} {choices:>5} choices")
    print(f"{'Cumulative branches':<38} {running_total:>5,}")
    print()

print("-" * 70)
print(f"TOTAL FULL HOUSE HANDS: {running_total:,}")

print("\nFINAL COMPARISON")
print("-" * 70)

total_hands = comb(52, 5)

print(f"Favorable hands: {running_total:,}")
print(f"All possible hands: {total_hands:,}")
print(f"Probability: {running_total:,} / {total_hands:,}")
print(f"Probability: {running_total / total_hands:.8f}")
print("=" * 70)
```

```python
from math import comb

# ============================================================
# EXPERIMENT — CHANGE THESE PARAMETERS
# ============================================================

number_of_ranks = 5       # Allowed: 2 to 20
suits_per_rank =   4       # Allowed: 3 to 8

# Full house structure remains fixed
three_of_a_kind = 3
pair_size = 2
hand_size = three_of_a_kind + pair_size

# ============================================================
# 1. VALIDATE PARAMETERS
# ============================================================

if not isinstance(number_of_ranks, int) or not 2 <= number_of_ranks <= 20:
    raise ValueError("Number of ranks must be an integer between 2 and 20.")

if not isinstance(suits_per_rank, int) or not 3 <= suits_per_rank <= 8:
    raise ValueError("Suits per rank must be an integer between 3 and 8.")

# ============================================================
# 2. CALCULATE THE NUMBER OF HANDS
# ============================================================

total_cards = number_of_ranks * suits_per_rank

total_hands = comb(total_cards, hand_size)

# Choose the triple rank, its suits,
# another rank, and its suits.
full_house_hands = (
    number_of_ranks
    * comb(suits_per_rank, three_of_a_kind)
    * (number_of_ranks - 1)
    * comb(suits_per_rank, pair_size)
)

# ============================================================
# 3. CALCULATE PROBABILITY
# ============================================================

probability = full_house_hands / total_hands

# ============================================================
# 4. DISPLAY RESULTS
# ============================================================

print("=" * 70)
print("FULL HOUSE — PARAMETER EXPERIMENT")
print("=" * 70)

print("\nCUSTOM DECK")
print("-" * 70)
print(f"Number of ranks: {number_of_ranks}")
print(f"Suits per rank: {suits_per_rank}")
print(f"Total cards: {number_of_ranks} x {suits_per_rank} = {total_cards}")

print("\nALL POSSIBLE HANDS")
print("-" * 70)
print(f"C({total_cards}, 5) = {total_hands:,}")

print("\nFULL HOUSE HANDS")
print("-" * 70)
print(
    f"{number_of_ranks} x C({suits_per_rank}, 3) "
    f"x {number_of_ranks - 1} x C({suits_per_rank}, 2)"
)
print(f"= {full_house_hands:,}")

print("\nPROBABILITY")
print("-" * 70)
print(f"P(full house) = {full_house_hands:,} / {total_hands:,}")
print(f"              = {probability:.10f}")
print(f"              = {probability * 100:.6f}%")

print("\n" + "=" * 70)
```
