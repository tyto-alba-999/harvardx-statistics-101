### Example 1.4.15: Permutations of a word

How many ways are there to permute the letters in the word LALALAAA? To determine a permutation, we just need to choose where the 5 A's go, in the 8 available slots (or, equivalently, choose where the 3 L's go). So there are

$$\binom{8}{5} = \binom{8}{3} = \frac{8 \cdot 7 \cdot 6}{3!} = 56 \text{ permutations}.$$

---

### Understanding the example

The word LALALAAA contains 8 letters:

- 5 identical A's.
- 3 identical L's.

Since identical letters cannot be distinguished from one another, exchanging two A's or two L's does not produce a new arrangement.

We can therefore determine a permutation by choosing the positions occupied by the A's.

There are 8 available positions, and we must select 5 of them:

$$\binom{8}{5} = \frac{8!}{5!3!} = 56$$

Alternatively, we can choose the 3 positions occupied by the L's:

$$\binom{8}{3} = \frac{8!}{3!5!} = 56$$

Both approaches give the same result because choosing the positions of the A's automatically determines the positions of the L's, and vice versa.

**Key idea:** When a word contains repeated letters, we must avoid counting arrangements that differ only by exchanging identical letters.

```python
from collections import Counter
from math import factorial, comb
from itertools import combinations

# ============================================================
# WORD
# ============================================================

word = "LALALAAA"

# ============================================================
# COUNT THE LETTERS
# ============================================================

letter_counts = Counter(word)

n = len(word)
a_count = letter_counts["A"]
l_count = letter_counts["L"]

# ============================================================
# CALCULATIONS
# ============================================================

total_permutations = factorial(n)

for count in letter_counts.values():
    total_permutations //= factorial(count)

# Choosing the positions of A or L
choose_A = comb(n, a_count)
choose_L = comb(n, l_count)

# ============================================================
# DISPLAY RESULTS
# ============================================================

print("=" * 65)
print("EXAMPLE 1.4.15 — PERMUTATIONS OF A WORD")
print("=" * 65)

print(f"\nOriginal word: {word}")
print(f"Total number of letters: {n}")

print("\nLETTER FREQUENCIES")
print("-" * 65)

for letter, count in letter_counts.items():
    print(f"Letter {letter}: {count} occurrences")

print("\nMETHOD 1 — CHOOSE THE POSITIONS OF A")
print("-" * 65)

print(f"Number of available positions: {n}")
print(f"Number of A's to place: {a_count}")

print(f"\nC({n}, {a_count}) = {n}! / ({a_count}! x {l_count}!)")
print(f"                = {factorial(n)} / "
      f"({factorial(a_count)} x {factorial(l_count)})")
print(f"                = {choose_A}")

print("\nMETHOD 2 — CHOOSE THE POSITIONS OF L")
print("-" * 65)

print(f"C({n}, {l_count}) = {n}! / ({l_count}! x {a_count}!)")
print(f"                = {factorial(n)} / "
      f"({factorial(l_count)} x {factorial(a_count)})")
print(f"                = {choose_L}")

print("\nVERIFICATION")
print("-" * 65)

print(f"Total arrangements if all letters were distinct: {n}! = {total_permutations}")
print(f"Number of arrangements of identical A's and L's: "
      f"{factorial(a_count)} x {factorial(l_count)}")
print(f"Distinct permutations: {factorial(n)} / "
      f"({factorial(a_count)} x {factorial(l_count)})")
print(f"                    = {total_permutations // "
      f"(factorial(a_count) * factorial(l_count))}")

print(f"\nResult using A positions: {choose_A}")
print(f"Result using L positions: {choose_L}")

print("\nVerification:",
      "PASSED" if choose_A == choose_L == total_permutations //
      (factorial(a_count) * factorial(l_count)) else "FAILED")

print("\n" + "=" * 65)
```
