### Example 1.4.15: Permutations of a word

How many ways are there to permute the letters in the word LALALAAA? To determine a permutation, we just need to choose where the 5 A's go, in the 8 available slots (or, equivalently, choose where the 3 L's go). So there are

$$\binom{8}{5} = \binom{8}{3} = \frac{8 \cdot 7 \cdot 6}{3!} = 56 \text{ permutations}.$$

---

### Understanding the example

The word LALALAAA contains 8 letters:

- 5 identical A's.
- 3 identical L's.

Since identical letters cannot be distinguished from one another, exchanging two A's or two L's does not produce a new arrangement.

We can therefore determine a permutation by choosing the positions occupied by the A's.

There are 8 available positions, and we must select 5 of them:

$$\binom{8}{5} = \frac{8!}{5!3!} = 56$$

Alternatively, we can choose the 3 positions occupied by the L's:

$$\binom{8}{3} = \frac{8!}{3!5!} = 56$$

Both approaches give the same result because choosing the positions of the A's automatically determines the positions of the L's, and vice versa.

**Key idea:** When a word contains repeated letters, we must avoid counting arrangements that differ only by exchanging identical letters.

```python
from collections import Counter
from math import factorial, comb

# ============================================================
# WORD
# ============================================================

word = "LALALAAA"

# ============================================================
# COUNT THE LETTERS
# ============================================================

letter_counts = Counter(word)

n = len(word)
a_count = letter_counts["A"]
l_count = letter_counts["L"]

# ============================================================
# CALCULATIONS
# ============================================================

total_arrangements = factorial(n)
repeated_letter_arrangements = factorial(a_count) * factorial(l_count)

distinct_permutations = total_arrangements // repeated_letter_arrangements

choose_A = comb(n, a_count)
choose_L = comb(n, l_count)

# ============================================================
# DISPLAY RESULTS
# ============================================================

print("=" * 65)
print("PERMUTATIONS OF A WORD")
print("=" * 65)

print(f"\nOriginal word: {word}")
print(f"Total number of letters: {n}")

print("\nLETTER FREQUENCIES")
print("-" * 65)

for letter, count in letter_counts.items():
    print(f"Letter {letter}: {count} occurrences")

print("\nMETHOD 1 — CHOOSE THE POSITIONS OF A")
print("-" * 65)

print(f"Number of available positions: {n}")
print(f"Number of A's to place: {a_count}")
print(f"\nC({n}, {a_count}) = {n}! / ({a_count}! x {l_count}!)")
print(f"                = {factorial(n)} / "
      f"({factorial(a_count)} x {factorial(l_count)})")
print(f"                = {factorial(n)} / {repeated_letter_arrangements}")
print(f"                = {choose_A}")

print("\nMETHOD 2 — CHOOSE THE POSITIONS OF L")
print("-" * 65)

print(f"C({n}, {l_count}) = {n}! / ({l_count}! x {a_count}!)")
print(f"                = {factorial(n)} / "
      f"({factorial(l_count)} x {factorial(a_count)})")
print(f"                = {factorial(n)} / {repeated_letter_arrangements}")
print(f"                = {choose_L}")

print("\nVERIFICATION")
print("-" * 65)

print(f"Total arrangements if all letters were distinct: {n}! = {total_arrangements}")

print(f"Arrangements of identical letters:")
print(f"  {a_count}! x {l_count}! = {repeated_letter_arrangements}")

print("\nDistinct permutations:")
print(f"  {factorial(n)} / {repeated_letter_arrangements} = {distinct_permutations}")

print(f"\nResult using A positions: {choose_A}")
print(f"Result using L positions: {choose_L}")

print(
    "\nVerification:",
    "PASSED" if choose_A == choose_L == distinct_permutations else "FAILED"
)

print("\n" + "=" * 65)
```

```python
from itertools import combinations

# ============================================================
# WORD SETUP
# ============================================================

word = "LALALAAA"

n = len(word)
a_count = word.count("A")
l_count = word.count("L")

positions = range(1, n + 1)

# ============================================================
# GENERATE ALL POSSIBLE ARRANGEMENTS
# ============================================================

a_positions_list = list(combinations(positions, a_count))

print("=" * 75)
print("VISUALIZING THE PERMUTATIONS OF LALALAAA")
print("=" * 75)

print("\nSTEP 1 — The eight available positions")
print("-" * 75)

print("Positions: ", "  ".join(str(i) for i in positions))

print("\nSTEP 2 — Choose five positions for the letter A")
print("-" * 75)

print(f"Number of ways to choose these positions: C({n}, {a_count})")
print(f"                  = {len(a_positions_list)}")

print("\nSTEP 3 — Generate all distinct arrangements")
print("-" * 75)

for i, a_positions in enumerate(a_positions_list, start=1):

    arrangement = [
        "A" if position in a_positions else "L"
        for position in positions
    ]

    word_result = "".join(arrangement)

    print(f"{i:2}. {word_result}   A positions: {a_positions}")

print("\nSTEP 4 — Final result")
print("-" * 75)

print(f"Total possible arrangements: {len(a_positions_list)}")
print(f"Binomial coefficient: C(8, 5) = 56")

print("\nCONCLUSION")
print("-" * 75)

print(
    "Every choice of five positions for A uniquely determines "
    "the remaining three positions for L."
)

print(
    "Therefore, the number of distinct permutations is "
    "C(8, 5) = 56."
)
```

```python
from collections import Counter
from math import factorial

# ============================================================
# GENERAL PERMUTATIONS OF A WORD — TEXT VISUALIZATION
# ============================================================

# EDIT THIS VALUE
word = "GHOSTTICK001"

# Maximum allowed length
MAX_LENGTH = 12

print("=" * 75)
print("GENERAL PERMUTATIONS OF A WORD")
print("=" * 75)

# ------------------------------------------------------------
# 1. VALIDATE THE INPUT
# ------------------------------------------------------------

if not isinstance(word, str):
    print("ERROR: The input must be a string.")

elif len(word) == 0:
    print("ERROR: The word cannot be empty.")

elif len(word) > MAX_LENGTH:
    print(f"ERROR: The maximum allowed length is {MAX_LENGTH} characters.")
    print(f"Your input contains {len(word)} characters.")

else:
    # --------------------------------------------------------
    # 2. COUNT CHARACTERS
    # --------------------------------------------------------

    counts = Counter(word)
    n = len(word)

    print(f"\nInput: {repr(word)}")
    print(f"Length: {n} characters")

    print("\nCHARACTER FREQUENCIES")
    print("-" * 75)

    for character, count in counts.items():
        label = repr(character)
        print(f"{label:12} appears {count} time(s)")

    # --------------------------------------------------------
    # 3. CALCULATE DISTINCT PERMUTATIONS
    # --------------------------------------------------------

    denominator = 1

    for count in counts.values():
        denominator *= factorial(count)

    total = factorial(n)
    distinct = total // denominator

    print("\nMATHEMATICAL CALCULATION")
    print("-" * 75)

    print(f"Total characters: {n}")
    print(f"All possible arrangements if every character were distinct: {n}!")

    print(f"\nNumerator: {n}! = {total}")

    print("\nDenominator: product of the factorials of character frequencies")

    denominator_parts = [
        f"{count}!" for count in counts.values()
    ]

    print(" x ".join(denominator_parts), "=", denominator)

    print("\nNumber of distinct permutations:")
    print(f"{total} / {denominator} = {distinct}")

    # --------------------------------------------------------
    # 4. TEXT VISUALIZATION
    # --------------------------------------------------------

    print("\nHOW REPEATED CHARACTERS AFFECT THE COUNT")
    print("-" * 75)

    for character, count in counts.items():
        print(
            f"{repr(character)} occurs {count} time(s): "
            f"divide by {count}! = {factorial(count)}"
        )

    print("\nFINAL RESULT")
    print("-" * 75)
    print(f"The sequence has {distinct:,} distinct permutations.")

    print("\n" + "=" * 75)
```

### General Formula (Permutations of a String with Repeated Characters)

For a string containing $n$ characters, the number of distinct permutations is:

$$
\boxed{P = \frac{n!}{m_1! \times m_2! \times \cdots \times m_r!}}
$$

Where:

- $n$ is the total number of characters in the string, including spaces, digits, and punctuation.
- $r$ is the number of distinct characters.
- $m_i$ is the frequency (number of occurrences) of the $i$-th distinct character.

The formula accounts for repeated characters, which are indistinguishable when calculating distinct permutations.
