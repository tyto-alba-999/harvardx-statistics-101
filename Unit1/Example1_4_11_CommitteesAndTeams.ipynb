
### Example 1.4.11 (Committees and teams)

Consider a group of four people.

* **(a)** How many ways are there to choose a two-person committee?
* **(b)** How many ways are there to break the people into two teams of two?

---

### Solution

#### **(a)** Choosing a two-person committee

One way to count the possibilities is by directly listing them out. Labeling the people as $1, 2, 3, 4$, the possible committees are:

$$
\{1, 2\}, \{1, 3\}, \{1, 4\}, \{2, 3\}, \{2, 4\}, \{3, 4\}
$$

*(This gives a total of 6 possibilities).*

Another approach is to use the **multiplication rule** combined with an **adjustment for overcounting**:

* There are $4$ ways to choose the first person on the committee.
* There are $3$ ways to choose the second person.

However, this ordered process counts each committee twice. For example, picking person $1$ then person $2$ results in the exact same committee as picking person $2$ then person $1$ ($\{1, 2\} = \{2, 1\}$). Since we have overcounted by a factor of $2$, we divide by $2$ to get the correct count:

$$
\frac{4 \times 3}{2} = 6
$$

---

#### **(b)** Breaking into two teams of two

To find the number of ways to break the four people into two separate teams of two, let's look at a few ways to think about it:

1. **Direct Listing:** Labeling the people as $1, 2, 3, 4$, we can write out the possible pairings:

   * Team 1: $\{1, 2\}$ and Team 2: $\{3, 4\}$
   * Team 1: $\{1, 3\}$ and Team 2: $\{2, 4\}$
   * Team 1: $\{1, 4\}$ and Team 2: $\{2, 3\}$

   *(This gives a total of 3 distinct ways).*

2. **Fixing a Teammate:** Alternatively, you can simply specify who is on person $1$'s team (which automatically determines the other team). Once person $1$ is paired with one of the remaining $3$ people, the other two people automatically form the second team.

3. **Using Part (a) with Adjustment:** We can use our result from part (a), which showed there are $6$ ways to choose a single two-person committee. However, choosing one team of two automatically leaves behind the second team of two. This overcounts the overall division into *two unnamed teams* by a factor of $2$, because picking team $\{1, 2\}$ (leaving $\{3, 4\}$) is the same partition as picking team $\{3, 4\}$ (leaving $\{1, 2\}$). Adjusting for this overcounting gives:

$$
\frac{6}{2} = 3
$$


```python

from itertools import combinations

# Four distinct people
people = (1, 2, 3, 4)

# --------------------------------------------------
# Part (a): Choosing a two-person committee
# --------------------------------------------------

# Method 1: Direct listing
committees = list(combinations(people, 2))

print("PART (a): TWO-PERSON COMMITTEES")
print("-" * 38)

for committee in committees:
    print(committee)

print(f"\nTotal committees: {len(committees)}")

# Method 2: Multiplication rule with overcounting
ordered_selections = [
    (first, second)
    for first in people
    for second in people
    if first != second
]

print(f"Ordered selections: {len(ordered_selections)}")
print(f"After correcting for overcounting: "
      f"{len(ordered_selections)} / 2 = "
      f"{len(ordered_selections) // 2}")


# --------------------------------------------------
# Part (b): Dividing into two unnamed teams of two
# --------------------------------------------------

print("\nPART (b): DIVISIONS INTO TWO TEAMS")
print("-" * 38)

# Method 1: Direct listing
divisions = []

for team1 in combinations(people, 2):
    team2 = tuple(p for p in people if p not in team1)

    # Keep only divisions where person 1 belongs to team 1.
    # This avoids counting a division twice.
    if 1 in team1:
        divisions.append((team1, team2))

print("\nMethod 1 — Direct listing:")
for team1, team2 in divisions:
    print(f"Team 1: {team1} | Team 2: {team2}")

print(f"Total divisions: {len(divisions)}")

# Method 2: Fixing person 1's teammate
print("\nMethod 2 — Fixing person 1's teammate:")
for teammate in people[1:]:
    team1 = (1, teammate)
    team2 = tuple(p for p in people if p not in team1)
    print(f"Person 1 with {teammate}: "
          f"{team1} | {team2}")

# Method 3: Using part (a) with adjustment
print("\nMethod 3 — Correcting for overcounting:")
print(f"{len(committees)} / 2 = "
      f"{len(committees) // 2}")

# Verify the results
assert len(committees) == 6
assert len(divisions) == 3
```

```python

print("COMMITTEES")
print("=" * 42)
print("Each bracketed pair is one possible committee.\n")

for i, committee in enumerate(committees, start=1):
    selected = set(committee)
    representation = "  ".join(
        f"[{person}]" if person in selected else f" {person} "
        for person in people
    )
    print(f"{i}. {representation}")

print("\nTEAM DIVISIONS")
print("=" * 42)
print("Each line represents one complete division.\n")

for i, (team1, team2) in enumerate(divisions, start=1):
    print(f"Division {i}")
    print(f"    Team 1: {team1}")
    print(f"    Team 2: {team2}")
    print("    " + "-" * 24)

print("FINAL COUNT")
print("-" * 24)
print(f"Two-person committees: {len(committees)}")
print(f"Divisions into two teams: {len(divisions)}")
```

### Definition 1.4.12 (Binomial coefficient)

For any nonnegative integers $k$ and $n$, the binomial coefficient $\binom{n}{k}$, read as "$n$ choose $k$", is the number of subsets of size $k$ for a set of size $n$.

For example, $\binom{4}{2} = 6$, as shown in Example 1.4.11. The binomial coefficient $\binom{n}{k}$ is sometimes called a *combination*, but we do not use that terminology here since "combination" is such a useful general-purpose word. Algebraically, binomial coefficients can be computed as follows.

---

### Theorem 1.4.13 (Binomial coefficient formula)

* For $k \le n$, we have:

$$\binom{n}{k} = \frac{n(n-1)\cdots(n-k+1)}{k!} = \frac{n!}{(n-k)!k!}$$

* For $k > n$, we have $\binom{n}{k} = 0$.

---

### Proof

Let $S$ be a set with $\vert{}S\vert{} = n$. Any subset of $S$ has size at most $n$, so $\binom{n}{k} = 0$ for $k > n$.

Now let $0 \le k \le n$. By Theorem 1.4.6, there are $n(n-1)\cdots(n-k+1)$ ways to make an *ordered* choice of $k$ elements from $S$ without replacement. This ordered process overcounts each subset of interest by a factor of $k!$ (since there are $k!$ ways to order any chosen subset of $k$ elements, and we don't care how these elements are ordered).

Thus, we obtain the correct count for the number of unordered subsets by dividing the number of ordered choices by $k!$:

$$\binom{n}{k} = \frac{n(n-1)\cdots(n-k+1)}{k!} = \frac{n!}{(n-k)!k!}$$

---

### Example: Understanding the proof with a set of four elements

Let $S = \{1,2,3,4\}$ and suppose we want to select two elements.

First, count the ordered selections without replacement:

$$4 \times 3 = 12$$

For every subset of size two, there are $2! = 2$ possible orders. For example, the subset $\{1,2\}$ can be obtained as $(1,2)$ or $(2,1)$.

Since the order does not matter, divide by $2!$:

$$\binom{4}{2} = \frac{4 \times 3}{2!} = \frac{12}{2} = 6$$

The six subsets are:

$$\{1,2\}, \{1,3\}, \{1,4\}, \{2,3\}, \{2,4\}, \{3,4\}$$

This illustrates why we divide the number of ordered selections by $k!$: every subset is counted exactly $k!$ times when order is considered.

```python
from math import factorial
from itertools import combinations

def binomial_coefficient(n, k):
    if k > n:
        return 0
    return factorial(n) // (factorial(k) * factorial(n - k))


print("=" * 65)
print("VERIFICATION OF THE BINOMIAL COEFFICIENT FORMULA")
print("=" * 65)

test_cases = [(4, 2), (5, 2), (5, 3), (6, 0), (6, 6), (3, 5)]

for n, k in test_cases:
    result = binomial_coefficient(n, k)

    if k <= n:
        numerator = factorial(n)
        denominator = factorial(n - k) * factorial(k)

        print(f"\nChoosing {k} elements from a set of {n}:")
        print(f"  Formula: C({n}, {k}) = {n}! / [({n}-{k})! × {k}!]")
        print(f"          = {numerator} / {denominator}")
        print(f"  Number of possible subsets: {result}")
    else:
        print(f"\nChoosing {k} elements from a set of {n}:")
        print(f"  Since {k} > {n}, no such subset exists.")
        print(f"  Therefore, C({n}, {k}) = {result}")


print("\n" + "=" * 65)
print("DIRECT VERIFICATION: C(4, 2)")
print("=" * 65)

S = {1, 2, 3, 4}
subsets = list(combinations(S, 2))

print("The set is S = {1, 2, 3, 4}.")
print("All possible subsets containing exactly two elements are:\n")

for i, subset in enumerate(subsets, start=1):
    print(f"  {i}. {{{subset[0]}, {subset[1]}}}")

print(f"\nTotal subsets found by enumeration: {len(subsets)}")
print(f"Total subsets given by the formula: {binomial_coefficient(4, 2)}")
print("Verification:", "PASSED" if len(subsets) == binomial_coefficient(4, 2) else "FAILED")
```

```python
from itertools import permutations

S = (1, 2, 3, 4)
k = 2

ordered_selections = list(permutations(S, k))
subsets = list(combinations(S, k))

print("=" * 65)
print("WHY DO WE DIVIDE BY k!?")
print("=" * 65)

print("\nSTEP 1 — Count ordered selections")
print("----------------------------------")

for i, selection in enumerate(ordered_selections, start=1):
    print(f"{i:2}. ({selection[0]}, {selection[1]})")

print(f"\nTotal ordered selections: {len(ordered_selections)}")

print("\nSTEP 2 — Group selections representing the same subset")
print("-------------------------------------------------------")

for i, subset in enumerate(subsets, start=1):
    orders = list(permutations(subset))
    formatted_orders = " and ".join(
        f"({a}, {b})" for a, b in orders
    )

    print(f"{i}. Subset {{{subset[0]}, {subset[1]}}}: {formatted_orders}")

print("\nSTEP 3 — Remove the overcounting")
print("--------------------------------")

print(f"Each subset is counted {factorial(k)} times.")
print(f"Number of ordered selections: {len(ordered_selections)}")
print(f"Number of orders per subset: {factorial(k)}")

print(
    f"\nNumber of distinct subsets = "
    f"{len(ordered_selections)} / {factorial(k)} "
    f"= {len(subsets)}"
)

print("\nCONCLUSION")
print("----------")
print(
    "Dividing by k! removes the different orderings of the same "
    "subset, leaving each subset counted exactly once."
)
```

```python
from itertools import permutations, combinations
from math import factorial

S = (1, 2, 3, 4)
k = 3

ordered_selections = list(permutations(S, k))
subsets = list(combinations(S, k))

print("=" * 75)
print("BINOMIAL COEFFICIENT: UNDERSTANDING THE PROOF WHEN n= 4 AND k = 3")
print("=" * 75)

print("\nSTEP 1 — Count all ordered selections")
print("--------------------------------------")

for i, selection in enumerate(ordered_selections, start=1):
    print(f"{i:2}. ({selection[0]}, {selection[1]}, {selection[2]})")

print(f"\nTotal ordered selections: {len(ordered_selections)}")
print("Calculation: 4 × 3 × 2 = 24")

print("\nSTEP 2 — Group selections representing the same subset")
print("-------------------------------------------------------")

for i, subset in enumerate(subsets, start=1):
    orders = list(permutations(subset))

    print(f"\nSubset {i}: {{{subset[0]}, {subset[1]}, {subset[2]}}}")

    for order in orders:
        print(f"    ({order[0]}, {order[1]}, {order[2]})")

    print(f"    Number of orderings: {len(orders)}")

print("\nSTEP 3 — Remove the overcounting")
print("--------------------------------")

print(f"Each subset is counted 3! = {factorial(k)} times.")
print(f"Total ordered selections: {len(ordered_selections)}")
print(f"Number of orderings per subset: {factorial(k)}")

print(
    f"\nNumber of distinct subsets = "
    f"{len(ordered_selections)} / {factorial(k)} "
    f"= {len(subsets)}"
)

print("\nSTEP 4 — Verify using the binomial coefficient formula")
print("------------------------------------------------------")

print("C(4, 3) = 4! / [(4 - 3)! × 3!]")
print(f"       = 24 / (1 × 6)")
print(f"       = {len(subsets)}")

print("\nCONCLUSION")
print("----------")
print(
    "There are 24 ordered selections, but every subset of "
    "three elements appears in 6 different orders."
)
print(
    "Dividing by 3! = 6 gives the 4 distinct subsets, "
    "counting each one exactly once."
)
```

```python
from itertools import combinations
from math import factorial

# ============================================================
# CHOOSE YOUR VALUES HERE
# ============================================================

n = 5  # Set size (choose a value between 0 and 10)
k = 3  # Subset size (choose any nonnegative integer)

# ============================================================
# CALCULATIONS
# ============================================================

print("=" * 65)
print("EXPLORING BINOMIAL COEFFICIENTS")
print("=" * 65)

print(f"\nYou chose n = {n} and k = {k}.")

if not (0 <= n <= 10) or k < 0:
    print("\nInvalid values. Choose 0 <= n <= 10 and k >= 0.")

elif k > n:
    print(f"\nSince k = {k} > n = {n}, no such subset exists.")
    print(f"Therefore, C({n}, {k}) = 0.")

else:
    result = factorial(n) // (factorial(k) * factorial(n-k))
    ordered_count = factorial(n) // factorial(n-k)

    # --------------------------------------------------------
    # BINOMIAL COEFFICIENT
    # --------------------------------------------------------

    print("\nBINOMIAL COEFFICIENT")
    print("-" * 65)

    print(f"C({n}, {k}) = {n}! / [({n}-{k})! x {k}!]")
    print(f"           = {factorial(n)} / "
          f"({factorial(n-k)} x {factorial(k)})")
    print(f"           = {factorial(n)} / "
          f"{factorial(n-k) * factorial(k)}")
    print(f"           = {result}")

    # --------------------------------------------------------
    # COUNTING ORDERED SELECTIONS
    # --------------------------------------------------------

    print("\nCOUNTING ORDERED SELECTIONS")
    print("-" * 65)

    if k == 0:
        print("There is exactly one ordered selection: the empty selection.")
    else:
        factors = " x ".join(str(n-i) for i in range(k))
        print(f"Number of ordered selections:")
        print(f"  {factors} = {ordered_count}")

    print(f"\nNumber of orderings per subset:")
    print(f"  {k}! = {factorial(k)}")

    print(f"\nNumber of unordered subsets:")
    print(f"  {ordered_count} / {factorial(k)} = {result}")

    # --------------------------------------------------------
    # ALL POSSIBLE SUBSETS
    # --------------------------------------------------------

    print("\nALL POSSIBLE SUBSETS")
    print("-" * 65)

    subsets = list(combinations(range(1, n+1), k))

    for i, subset in enumerate(subsets, start=1):
        print(f"{i:3}. {set(subset)}")

    print(f"\nTotal number of subsets: {len(subsets)}")

    # --------------------------------------------------------
    # VERIFICATION
    # --------------------------------------------------------

    print("\nVERIFICATION")
    print("-" * 65)

    print(f"Ordered selections / k! = {ordered_count} / {factorial(k)}")
    print(f"                       = {result}")
    print(f"Subsets found by enumeration = {len(subsets)}")

    print(
        "\nVerification: PASSED"
        if len(subsets) == result
        else "\nVerification: FAILED"
    )

print("\n" + "=" * 65)
```
