### Example 1.4.3: Ice Cream Cones

An ice cream cone is determined by two choices:

- **Cone type:** cake or waffle
- **Flavor:** chocolate, vanilla, or strawberry

By the multiplication rule, if the first choice has $m$ possibilities and the second has $n$ possibilities for every first choice, then the total number of outcomes is

$$
|\Omega| = m \times n.
$$

In this example, there are $2$ cone types and $3$ flavors. Therefore,

$$
|\Omega| = 2 \times 3 = 6.
$$

The order in which the choices are made does not change the number of possible outcomes.

**Important:** The multiplication rule applies here because each cone type can be paired with all three flavors. If chocolate were forbidden with a waffle cone, there would be only $5$ possible outcomes.

```python
from itertools import product

# Define the possible choices
cone_types = ["Cake", "Waffle"]
flavors = ["Chocolate", "Vanilla", "Strawberry"]

# Enumerate all possible combinations
outcomes = list(product(cone_types, flavors))

# Display the sample space
print("All possible ice cream cones:")
for i, (cone, flavor) in enumerate(outcomes, start=1):
    print(f"{i}. {flavor} ice cream in a {cone.lower()} cone")

# Verify the multiplication rule
print("\nNumber of cone types:", len(cone_types))
print("Number of flavors:", len(flavors))
print("Number of possible outcomes:", len(outcomes))
print("Multiplication rule:", len(cone_types), "*", len(flavors),
      "=", len(cone_types) * len(flavors))

assert len(outcomes) == len(cone_types) * len(flavors) == 6
```

```python
cone_types = ["Cake cone", "Waffle cone"]
flavors = ["Chocolate", "Vanilla", "Strawberry"]

print("Ice cream")
for i, cone in enumerate(cone_types):
    print("├──" if i == 0 else "└──", cone)
    branch = "│   " if i == 0 else "    "

    for j, flavor in enumerate(flavors):
        connector = "├──" if j < len(flavors) - 1 else "└──"
        print(f"{branch}{connector} {flavor}")

print("\nTotal possibilities:", len(cone_types) * len(flavors))
```

```python
# Chocolate ice cream is unavailable with a waffle cone
restricted_outcomes = [
    (cone, flavor)
    for cone, flavor in outcomes
    if not (cone == "Waffle" and flavor == "Chocolate")
]

print("Outcomes when waffle + chocolate is forbidden:")
for i, (cone, flavor) in enumerate(restricted_outcomes, start=1):
    print(f"{i}. {flavor} ice cream in a {cone.lower()} cone")

print("\nNumber of valid outcomes:", len(restricted_outcomes))

assert len(restricted_outcomes) == 5
```
