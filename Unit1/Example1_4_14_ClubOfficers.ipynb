### Example 1.4.14: Club officers

In a club with $n$ people, there are $n(n-1)(n-2)$ ways to choose a president, vice president, and treasurer, and there are $\binom{n}{3} = \frac{n(n-1)(n-2)}{3!}$ ways to choose $3$ officers without predetermined titles.

---

### Understanding the example

There are two different situations:

**1. Choosing officers with predetermined titles**

The three positions are different: president, vice president, and treasurer.

- There are $n$ choices for president.
- There are $n-1$ remaining choices for vice president.
- There are $n-2$ remaining choices for treasurer.

By the multiplication rule:

$$n(n-1)(n-2)$$

**2. Choosing three officers without predetermined titles**

Here, we only care about which three people are selected. Their order does not matter.

Therefore, we use the binomial coefficient:

$$\binom{n}{3} = \frac{n(n-1)(n-2)}{3!}$$

We divide by $3!$ because each group of three people can be arranged in $3! = 6$ different ways when assigning the three distinct positions.

**The key distinction:** When positions have different titles, order matters. When selecting a group without predetermined titles, order does not matter.

```python
from itertools import permutations, combinations
from math import factorial

# ============================================================
# CHOOSE THE NUMBER OF PEOPLE
# ============================================================

n = 5

# ============================================================
# CALCULATIONS
# ============================================================

print("=" * 70)
print("EXAMPLE 1.4.14 — CLUB OFFICERS")
print("=" * 70)

print(f"\nNumber of people in the club: n = {n}")

if n < 3:
    print("\nAt least 3 people are required to fill the three positions.")

else:
    people = tuple(range(1, n + 1))

    # Case 1: Distinct officer positions
    ordered_selections = list(permutations(people, 3))
    titled_count = len(ordered_selections)

    # Case 2: Three officers without predetermined titles
    groups = list(combinations(people, 3))
    untitled_count = len(groups)

    print("\nCASE 1 — OFFICERS WITH PREDETERMINED TITLES")
    print("-" * 70)

    print("President:")
    print(f"  {n} possible choices")

    print("Vice president:")
    print(f"  {n} - 1 = {n-1} possible choices")

    print("Treasurer:")
    print(f"  {n} - 2 = {n-2} possible choices")

    print("\nUsing the multiplication rule:")
    print(f"  {n} x {n-1} x {n-2} = {titled_count}")

    print(f"\nTotal possible assignments: {titled_count}")

    print("\nCASE 2 — THREE OFFICERS WITHOUT PREDETERMINED TITLES")
    print("-" * 70)

    print("Using the binomial coefficient formula:")

    print(f"  C({n}, 3) = {n}! / [({n}-3)! x 3!]")
    print(f"            = {factorial(n)} / "
          f"({factorial(n-3)} x {factorial(3)})")
    print(f"            = {factorial(n)} / "
          f"{factorial(n-3) * factorial(3)}")
    print(f"            = {untitled_count}")

    print(f"\nTotal possible groups: {untitled_count}")

    print("\nCOMPARISON")
    print("-" * 70)

    print(f"With predetermined titles: {titled_count}")
    print(f"Without predetermined titles: {untitled_count}")

    print(f"\nNumber of arrangements per group: 3! = {factorial(3)}")

    print(f"Verification:")
    print(f"  {titled_count} / {factorial(3)} = {titled_count // factorial(3)}")

    print(
        "\nVerification:",
        "PASSED" if titled_count == untitled_count * factorial(3)
        else "FAILED"
    )
```

```python
from itertools import permutations, combinations
from math import factorial

# ============================================================
# EXAMPLE SETUP
# ============================================================

people = (1, 2, 3, 4)

titles = ("President", "Vice President", "Treasurer")

# ============================================================
# VISUALIZATION
# ============================================================

print("=" * 75)
print("WHY DO OFFICER TITLES CHANGE THE NUMBER OF POSSIBILITIES?")
print("=" * 75)

print("\nTHE CLUB")
print("-" * 75)
print(f"Members: {people}")

groups = list(combinations(people, 3))

print("\nSTEP 1 — Select three people without predetermined titles")
print("-" * 75)

for i, group in enumerate(groups, start=1):
    print(f"Group {i}: {group}")

print(f"\nTotal groups: {len(groups)}")

print("\nSTEP 2 — Assign different titles to the first group")
print("-" * 75)

selected_group = groups[0]

print(f"Selected group: {selected_group}\n")

arrangements = list(permutations(selected_group))

for i, arrangement in enumerate(arrangements, start=1):
    print(f"Arrangement {i}:")
    for title, person in zip(titles, arrangement):
        print(f"    {title:<18}: Person {person}")
    print()

print(f"Total arrangements for this group: {len(arrangements)}")

print("\nSTEP 3 — Compare the two counting methods")
print("-" * 75)

print(f"Number of groups: {len(groups)}")
print(f"Arrangements per group: {factorial(3)}")
print(
    f"Total titled assignments: "
    f"{len(groups)} x {factorial(3)} = "
    f"{len(groups) * factorial(3)}"
)

print("\nCONCLUSION")
print("-" * 75)
print(
    "The same three people can occupy the three different "
    "positions in 6 different ways."
)
print(
    "Therefore, assigning titles produces 6 times as many "
    "possibilities as simply selecting a group of three."
)
```
